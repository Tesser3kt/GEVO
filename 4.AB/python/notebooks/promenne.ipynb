# Proměnné v Pythonu

O *proměnných* můžeme přemýšlet jako o "pojmenovaných úložištích údajů". Představme si, že v programu pracujeme s nějakým velmi dlouhým číslem nebo textem. Byla by velká otrava muset pořád dokola opisovat téže číslo/týž text pokaždé, kdy bychom s ním chtěli provést jakoukoli operaci.

Často se taky stává, že to ani nejde: náš program může dostávat informace z vnějšku, od uživatele skrz myš/klávesnici nebo z nějakého souboru uloženého na počítači. V obou případech si **chceme údaj**, se kterým pracujeme, **nějak pojmenovat** a odkazovat se na něj tímto jménem.

Jazykovou paralelou proměnných v Pythonu jsou *zájmena*. Uvažte, o kolik je příjemnější na vyjádření souvětí: "K zastávce přijel starý, polorozpadlý a celkově rezavý autobus a já do něj musel nastoupit," oproti "K zastávce přijel starý, polorozpadlý a celkově rezavý autobus a já do toho starého, polorozpadlého a celkově rezavého autobusu musel nastoupit."

Proměnné v Pythonu definujeme velmi jednoduše. Napíšeme zkrátka jméno proměnné a potom symbol `=`, na jehož pravé straně je objekt, který takto chceme pojmenovat. **S proměnnou pak můžeme pracovat úplně stejně jako s tím objektem** a proměnná vlastně "získá" jeho datový typ.

Několik příkladů:
- `number = 3` uloží do proměnné `number` celé číslo `3`. V ten moment má `number` datový typ `int` a tak je třeba s ním zacházet. Tedy, `number + 2` je v pořádku, ale třeba `number + "wtv"` nebo `number[:5]` ne.
- `number = 6.7` uloží do proměnné `number` desetinné číslo `6.7`.
- `word = "letadlo"` uloží do proměnné `word` string `"letadlo"`. Operace `word[2]` nebo `word + "autobus"` jsou zcela validní, ale třeba `word / 3` ne.

**Jména proměnných nemají nic společného ze stringy!** Python je nevnímá jako posloupnosti symbolů a mají datový typ podle toho, co je do nich uloženo. Obecně **nelze použít jakýkoliv symbol ve jméně proměnné**, je dobré se přidržet malých písmen bez diakritiky a čísel.

Pár pravidel pro jména proměnných:
- jména proměnných **nesmějí začínat číslem**;
- jména proměnných **nesmějí obsahovat mezeru**, ta se většinou nahrazuje symbolem `_` u proměnných, kde se hodí víceslovný název. Například proměnnou pro křestní jméno člověka můžeme pojmenovat `first_name`;
- jména proměnných nesmějí obsahovat mnoho speciálních znaků jako třeba `+ - * / % $ ( ) [ ] { }` a dalších.

Rozmyslete si, co je na konci programu uloženo v proměnné `x` v následujících příkladech.

## Shrnutí

Proměnné v Pythonu používáme buď pro pojmenování údajů, se kterými pak potřebujeme opakovaně pracovat nebo které dostaneme z vnějšku programu -- od uživatele nebo ze souboru apod. Jména proměnných mohou obsahovat malá i velká písmena, čísla a **některé** speciální znaky. Do proměnných ukládáme příkazem `x = ...`, kde `x` je název proměnné a `...` je jakýkoliv výraz, který Python umí spočítat.
