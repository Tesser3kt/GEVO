# Čísla v pythonu

Python rozlišuje dva typy čísel: *celá* a *desetinná*. Celá čísla se píšou, jak jsme zvyklí, zkrátka číslicemi a bez tečky. Například `23` nebo `-69` jsou celá čísla. Oproti tomu, desetinná čísla mají **vždy** tečku. Třeba `3.14` nebo `-0.331` jsou desetinná čísla, ale taky `1.0` nebo `0.0`. Z matematiky jste zvyklí, že číslo může být desetinné i celé, třeba číslo `4` mohu napsat jako `8/2`. **V Pythonu ne: každé číslo je buď celé, nebo desetinné.**

## Aritmetické operace
V podstatě všechny aritmetické operace, které znáte, zná i Python. S celými čísly lze dělat následující operace:
- sčítání: `2 + 3`
- odčítání: `4 - 3`
- násobení: `-3 * 7`
- **celočíselné dělení**: `16 // 3` (kolikrát se `3` vejde do `16`)
- **zbytek po dělení**: `16 % 3` (kolik zbude po dělení čísla `16` číslem `3`)
- mocnění: `6 ** 5`

**A co normální dělení?** Dělení beze zbytku *není operace na celých číslech*, protože může vyjít desetinné číslo. Třeba `14 / 4 = 3.5`. Když Python počítá podíl dvou celých čísel, tak je potají převede na desetinná a pak teprve provede výpočet. Tedy, když Python počítá `14 / 4`, tak nejprve z `14` udělá `14.0`, z `4` udělá `4.0` a potom spočítá `14.0 / 4.0`. **Dělení je operace na desetinných číslech!**

S desetinnými čísly lze dělat následující operace:
- sčítání: `4.3 + 6.6`
- odčítání: `-1.22 - 7.0`
- násobení: `22.1 * 7.1`
- dělení: `5.0 / 3.1`
- mocnění: `1.3 ** -7.1`

Co dělá Python, když napíšu třeba `2 * 3.7`. To samé, co při podílu dvou celých čísel. *Když je aspoň jedno číslo desetinné, Python provádí celou operaci v desetinných číslech.* Tedy, když napíšu `2 * 3.7`, tak Python udělá z celého čísla `2` desetinné číslo `2.0` a potom spočítá `2.0 * 3.7`.

## Shrnutí
Python uznává dva typy čísel: **celá** a **desetinná**. Celá čísla nemají nikdy tečku a desetinná ji vždy mají.
- Celá čísla lze sčítat (`+`), odčítat (`-`), násobit (`*`), celočíselně dělit (`//`), dělit se zbytkem (`%`) a mocnit (`**`).
- Desetinná čísla lze sčítat (`+`), odčítat (`-`), násobit (`*`), dělit (`/`) a mocnit (`**`).
- Když v jedné operace smíchám celá a desetinná čísla, celá operace *se provede v desetinných číslech*. Např. z `7 + 1.3` se stane `7.0 + 1.3`.
