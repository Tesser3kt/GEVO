# Datové typy v Pythonu

Každý údaj v Pythonu (číslo, slovo, cokoli jiného) má nějaký **typ**. Podle něho počítač ví, jak s tou věcí lze zacházet. Viděli jsme například, že čísla můžeme sčítat, odčítat, násobit atd., ale to samé nelze dělat ze stringy, které sice můžeme "sčítat" (slučování má stejný symbol jako sčítání čísel), ale odčítání nebo násobení stringů není definováno. Když se s typy pokusíme dělat něco, co nesmíme, Python ukáže chybu.

```python
"kočka" - "pes"
```

Tou chybou je `TypeError` čili "chyba typu". Zpráva za chybou osvětluje, co se stalo: `unsupported operand type(s) for -: 'str' and 'str'`, čili "`-` je nepovolená operace pro dva stringy".

*Typy* v Pythonu *mají svá jména*:
- celá čísla mají typ `int` (z anglického výrazu pro celá čísla, *integer*)
- desetinná čísla mají typ `float` (ze spojení *floating point*, což je název pro způsob representace desetinných čísel na počítači)
- stringy mají typ `str` (jako ... počkejte si ... *string*)

Jména typů neslouží jenom k jejich rozpoznání v chybových hláškách, ale také k převodu údajů mezi nimi. Python umí **za rozumných okolností** udělat ze stringu číslo a z čísla string. Všechna tři slova `int`, `float` a `str` jsou zároveň **funkce** (více o nich později), pomocí kterých mohu mezi datovými typy přecházet.

Když chci jeden datový typ převést na jiný, obalím ho do závorek `()` a napíši před něj název datového typu, do kterého ho chci převést, jako třeba `float("3.14")` nebo `str(221)`.

Převody mezi číselnýmy typy `int` a `float` fungují takto:
- `int(30)` nedělá nic, `30` už je celé číslo;
- `int(6.7)` vrátí celé číslo `6` (`int` **nezaokrouhluje** jenom zapomíná desetinnou část čísla);
- `float(30)` vrátí desetinné číslo `30.0` (pamatujte, že desetinná čísla v Pythonu mají vždy tečku);
- `float(6.7)` nedělá nic, `6.7` už je desetinné číslo.

Převody mezi stringy a čísly fungují takto:
- `str(30)` udělá z celého čísla `30` string `"30"`, tedy posloupnost symbolů `"3"` a `"0"`;
- `str(6.7)` udělá z desetinného čísla `6.7` string `"6.7"`, tedy posloupnost symbolů `"6"`, `"."` a `"7"`;
- `int("443")` udělá ze stringu `"443"` celé číslo `443`;
- `int("6.7")` dá chybu: string `"6.7"` odpovídá *desetinnému* a ne *celému* číslu;
- `int("kočka")` dá rovněž chybu: `"kočka"` není (v žádném rozšířeném filosofickém pojetí světa) celé číslo;
- `float("443")` udělá ze stringu `"443"` desetinné číslo `443.0`;
- `float("6.7")` udělá ze stringu `"6.7"` desetinné číslo `6.7`;
- `float("pes")` dá chybu: string `"pes"` si Python neumí vyložit jako desetinné číslo.

Vyzkoušejte si, že jste datové typy pochopili. V buňkách níže zkuste nejdřív výsledek určit, než je spustíte.

```python
int("60") / float("5")
```

```python
str(3.14)[-1]
```

```python
str(55) + str(21)
```
