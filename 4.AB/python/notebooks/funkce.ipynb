# Funkce v Pythonu

Funkce v Pythonu jsou vlastně takové malé programy uvnitř velkých programů. Slouží hlavně k tomu, aby mohl člověk nějaký postup jednoduše opakovat mnohokrát (často s jinými vstupními daty).

Uvažme příklad receptu na pečení bábovky: ten obsahuje postup přípravy těsta, který je téměř identický napříč mnoha recepty a ve své tradiční podobě obsahuje smíchání daných množství žloutků, mouky, másla a mléka (+ nějaká sůl, cukr...). Pochopitelně, množství surovin záleží na receptu a též na tom, kolik bábovky chci vůbec udělat.

V Pythonu bychom si mohli napsat funkci `udelej_testo(vejce, mouka, maslo, mleko)`, která by dostala čtyři vstupní údaje: počet vajec, hmotnost mouky, hmotnost másla a objem mléka. Uvnitř funkce (v jejím "těle") by pak byl postup přípravy těsta. Když bychom v programu napsali `udelej_testo(2, 150, 50, 100)`, tak by Python jednorázově nahradil vstupní údaje funkce `udelej_testo` těmito čísly a vypsal postup pro dělání těsta.

Funkce v Pythonu vyrobíme tak, že napíšeme klíčové slovo `def` následované jménem funkce (v příkladě výše `udelej_testo`) a pak v kulatých závorkách `()` čárkou oddělené údaje, které funkce potřebuje (v příkladě výše `(vejce, mouka, maslo, mleko)`). Po údajích následuje dvojtečka `:` a obsah funkce odsazený zleva. Funkce může jenom něco udělat (třeba spočítat, vytisknout atd.), ale může též odpovědět či **vrátit** nějakou hodnotu. Když chceme, aby se tak stalo, použijeme slovíčko `return`. **V ten moment funkce končí** a vrací hodnotu za tímto slovem. S tou pak můžeme dělat cokoliv: uložit ji do proměnné, vytisknout ji, použít ji k výpočtu atd.

Ukážeme si příklad funkce, která dostane jako vstupní data dvě čísla a vrátí jejich součet.

```python
def make_sum(a, b):
    return a + b
```

Vysvětlení:
- Po `def` následuje název funkce `make_sum`.
- Funkce `make_sum` potřebuje za vstupní údaje dvě čísla pojmenovaná `a` a `b`.
- Po slovu `return` následuje výraz `a + b`, takže ho Python spočítá a výsledek vrátí.

Když chceme funkci `make_sum` použít v programu, můžeme napsat například `make_sum(5, 3)`, která dá výsledek `8`.

```python
s = make_sum(5, 3)
print(s)
```

Pochopitelně, do funkce můžeme dosazovat i proměnné a vstupní data (též parametry) funkce budou nahrazeny jejich hodnotou v daný moment. Matoucí bývá, že proměnné v programu **se mohou jmenovat stejně jako parametry funkce.** Následující dva programy *dělají oba úplně totéž.*

```python
num1 = 3
num2 = 5

make_sum(num1, num2)
```

```python
a = 3
b = 5

make_sum(a, b)
```

Funkce mohou mít libovolný počet vstupních dat (třeba i `0` -- i v tomto případě se kulaté závorky vždy píšou) a mohou i nemusejí vracet nějakou hodnotu. Dole je příklad funkce, která pouze vytiskne string `"Hello, world!"` a nic jiného neudělá ani nevrátí.

```python
def greet_the_world():
    print("Hello, world!")
```

Tuto funkci použijeme stejně jako jakoukoli jinou. Kulaté závorky píšeme u funkcí vždy.

```python
greet_the_world()
```

Funkce mohou vracet libovolný datový typ. Dole je příklad funkce, která vrací `bool` a může být tímpádem použitá třeba v podmínce nebo ve `while` cyklu. Zjišťuje, jestli číslo končí na aspoň jednu `0`, nebo ne. Funkce vracející `bool` může rovnou vrátit třeba výsledek nějakého porovnání.

```python
def has_trailing_zero(number):
    return number % 10 == 0

numbers = [5, 6, -30, 20, 11, 45390, 2888999000, 3321, 8890, -22100]
for number in numbers:
    if has_trailing_zero(number):
        print(number)
```

### Datový typ `None`

Co si myslíte, že se stane v případě níže?

```python
x = greet_the_world()
print(x)
```

Vytiskne se věta `Hello, world!` a pak záhadné `None`. `None` je v Pythonu datový typ, který má jen jedinou hodnotu pojmenovanou stejně: `None`. Vlastně znamená "nic". **Pozor!** `None` není vůbec to samé, co `0`: číslo `0` má hodnotu a Python ho musí v paměti počítače ukládat. Hodnota `None` přímo znamená, že pod nějakým jménem (v tomto případě `x`) není uloženo vůbec nic.

Totiž, funkce `greet_the_world` nic nevrací, takže když se pokusíme uložit vrácenou hodnotu do nějaké proměnné, tak v ní nic nebude, což nám Python dává najevo tak, že tato proměnná dostane datový typ `None`. Můžeme si představovat, že každá funkce, která žádnou hodnotu nevrací, má jakoby na konci `return None`.

Typ `None` je užitečný na testování, jestli nám funkce něco vrátila, nebo ne. Představme si funkci, která dostane vstupní data `start` a `end` a vrátí nám seznam všech sudých čísel mezi `start` (včetně) a `end` (mimo). Rádi bychom oddělili případ, kdy mezi `start` a `end` zkrátka žádné sudé číslo není (třeba `start = 3` a `end = 4`) od případu, kdy jsou `start` a `end` zadány nesmyslně (třeba když `end < start`). V prvním případě vrátíme prostě prázdný seznam, ale v druhém nevrátíme nic (tedy `None`).

```python
def even_numbers_between(start, end):
    if end < start:
        # Když end < start, skonči a nevracej nic.
        return

    even_numbers = []
    for num in range(start, end):
        if num % 2 == 0:
            even_numbers.append(num)

    return even_numbers
```

Vysvětlení:
- Funkce `even_numbers_between` dostane dvě vstupní data: `start` a `end`.
- Na začátku je podmínka `end < start`, jejíž splnění způsobí `return`. Když za `return` nic nenapíšeme, znamená to totéž jako `return None`. V moment, kdy funkce narazí na slovo `return`, okamžitě končí a nic dalšího nedělá, takže řádky pod `return` se v případě `end < start` vůbec neprovedou.
- Když `end >= start` (tedy v opačném případě) vyrábíme prázdný seznam `even_numbers`. Pak pro každé číslo v rozsahu `<start, end)` testujeme, zda je sudé (dává zbytek `0` po dělení `2`). Když ano, tak ho přidáme do `even_numbers` přes funkci `list.append(x)`, která přidá `x` do seznamu `list` (té tečky `.` si zatím nevšímejte, vysvětlíme si ji později).
- Na konci cyklu vrátíme seznam `even_numbers`.

### Shrnutí

Funkce v Pythonu slouží k automatizaci příkazů, které chceme opakovat v programu vícekrát, nebo zkrátka k rozdělení programu do menších celků, jejichž význam je snadno pochopitelný. Definují se přes klíčové slovo `def`, za nímž následuje jméno funkce a pak v kulatých závorkách vstupní údaje. Funkce může mít libovolný počet vstupních údajů a může, ale nemusí, vracet nějakou hodnotu. Funkce, která nic nevrací, jakoby vrací `None`. V programu použijeme definovanou funkci tak, že napíšeme její jméno a v závorkách jí předáme vstupní údaje **v tom pořadí, v kterém jsou v její definici**.
