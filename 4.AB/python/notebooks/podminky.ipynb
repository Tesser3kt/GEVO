# Podmínky v Pythonu

Často potřebujeme, aby se náš program zachoval různě, podle toho jak dopadnou předchozí výpočty. Uvažme jednoduchý případ situace, kdy uživatel zadá svůj měsíc narození a my mu přiřadíme znamení zvěrokruhu, abychom uměli, zcela objektivně a nezaujatě, zhodnotit jeho charakter.

Těmto účelům slouží v Pythonu *podmínky*: výrazy, které vedou program různými směry, podle toho, jestli je nějaký výraz pravda, nebo lež.
