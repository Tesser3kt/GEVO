# Podmínky v Pythonu

Často potřebujeme, aby se náš program zachoval různě, podle toho jak dopadnou předchozí výpočty. Uvažme jednoduchý případ situace, kdy uživatel zadá měsíc a my mu vytiskneme jeho název v češtině.

Těmto účelům slouží v Pythonu *podmínky*: výrazy, které vedou program různými směry, podle toho, jestli je nějaký výraz pravda, nebo lež.

Program na výpis jména měsíce (z lenosti jen pro první tři měsíce) by mohl vypadat jako níže. Jsou v něm i funkce `print` a `input`, které jsme ještě neviděli. `print` slouží k vypisování dat z programu a `input` k získávání dat od uživatele.

```python
month = int(input("Zadejte měsíc narození: "))

if month == 1:
    print("Leden")
elif month == 2:
    print("Únor")
elif month == 3:
    print("Březen")
```

Podmínky v Pythonu píšeme ve tvaru `if vyraz:`, kde `vyraz` je cokoliv, co Python dokáže vyhodnotit jako pravdu nebo jako lež. V programu výše vidíme třeba výraz typu `month == 2`, který je pravda, když v proměnné `month` je uloženo číslo `2` a lež v opačném případě.
**Pozor!** Všimněte si dvou rovnítek ve výrazu `month == 2`. V Pythonu slouží **jedno rovnítko k ukládání do proměnných** a **dvě rovnítka k porovnávání dvou hodnot**.

Další věc, které si všimnout, je `:` na konci řádku `if month == 2:`. Podmínka v Pythonu **musí končit dvojtečkou** a následující **kód musí být odsazený zleva** (ideálně klávesou `Tab`). Tak Python ví, co se má vykonat pouze za této podmínky a co už naopak s podmínkou vůbec nesouvisí. Podívejte se na kód níže.

```python
number = 67
if number % 2 == 0:
    print("Číslo je sudé.")
print("Konec programu.")
```

Kód výše vytiskne `Konec programu.`, protože řádek `print("Konec programu.")` už **není odsazený zleva**, takže vůbec nesouvisí s podmínkou `if number % 2 == 0` (která znamená, že je číslo sudé -- dává zbytek 0 po dělení 2). Když do proměnné `number` uložíme třeba `40`, pak tento kód vytiskne `Číslo je sudé.` i `Konec programu.`

### Datový typ `bool`

Abychom přesně pochopili, které výrazy můžeme dát do podmínky `if vyraz:`, představíme si datový typ `bool` (pojmenovaný po angl. matematiku Georgi Booleovi). `bool` může obsahovat pouze dvě různé hodnoty: `True` (pravda) a `False` (lež). V proměnné `vyraz` v podmínce `if vyraz:` může být úplně cokoliv, co má datový typ `bool`.

Co myslíte, že program níže vypíše, pokud vůbec něco?

```python
2 == 3
```

### Logické operátory

Pár operátorů, pomocí kterých můžeme tvořit podmínky jsou
- `==`, `!=`: po řadě *rovná se* a *nerovná se*;
- `<`, `>`, `<=`, `>=`: po řadě *menší*, *větší*, *menší nebo rovno*, *větší nebo rovno*;
- `in`: operátor, který testuje, že jeden objekt je uvnitř druhého;
    - např. `"s" in "pes"` je `True` a `5 in [4, 7, 8.9]` je `False`.

Podmínky můžeme též slučovat pomocí logických spojek `and` a `or`.
- Spojka `and` vytvoří ze dvou podmínek novou podmínku, která je `True`, jenom když obě ze sloučených podmínek jsou `True`.
- Spojka `or` vytvoří ze dvou podmínek novou podmínku, která je `True`, když aspoň jedna (ale klidně obě) ze sloučených podmínek je `True`. 

Existuje též operátor `not`, který z `True` dělá `False` a z `False` dělá `True`. Třeba `not (3 < 5)` je `False`, protože `3 < 5` je `True`. Nejčastěji na něj narazíme ve spojce `not in`, která je pravdivá, když nějaký objekt *není* uvnitř druhého. Třeba `5 not in [4, 7, 8.9]` je `True`.

Zkuste si rozmyslet, jak dopadne kód níže. Závorky `()` určují, která logická operace se vyhodnotí jako první.

```python
name = "John"
number = 32

if ("n" in name or number < 30) and number >= 20:
    print("Yes.")
else:
    print("No.")
```

### `if`, `elif`, `else`

Ve svém původním programu na výpis názvu měsíce jsme použili ještě slovo `elif` a v kódu výše slovo `else`. Slovo `elif` je zkratka pro `else if` a značí podmínku, kterou má Python brát v úvahu **až v moment, kdy předchozí podmínka je `False`**. Mezi `if` a `elif` je velký rozdíl. Například program níže vytiskne jak `Číslo je větší než 10.`, tak `Číslo je větší než 20.`, protože obě podmínky jsou splněny.

```python
number = 25

if number > 10:
    print("Číslo je větší než 10.")
if number > 20:
    print("Číslo je větší než 20.")
```

Na druhou stranu, tento program vytiskne jenom `Číslo je větší než 10.`, protože Python hodnotí podmínku v `elif` až ve chvíli, kdy všechny předchozí podmínky byly `False`.

```python
number = 25

if number > 10:
    print("Číslo je větší než 10.")
elif number > 20:
    print("Číslo je větší než 20.")
```

Jak už možná tušíte, slovo `else` značí začátek kódu, který se má provést až v moment, kdy všechny podmínky nad ním byly `False`. Co myslíte, že vytiskne tento program?

```python
number = 5

if number > 10:
    print("Číslo je větší než 10.")
elif number > 20:
    print("Číslo je větší než 20.")
else:
    print("Tohle číslo je nějaké malé.")
```

### Noření podmínek

V Pythonu lze psát podmínky dovnitř podmínek. Někdy je tento přístup preferovanější než jejich slučování pomocí `and` a `or` pro jeho lepší čitelnost. Třeba tento program vytiskne `Jdu ven.`, když buď neprší, anebo prší a mám deštník.

```python
raining = False         # Zkuste změnit na True.
have_umbrella = True    # Zkuste změnit na False.

if raining:
    if have_umbrella:
        print("Jdu ven.")
    else:
        print("Zůstávám doma.")
else:
    print("Jdu ven.")
```

Stejný program zapsaný pomocí spojek by vypadal třeba takto. Je na vás rozhodnout se, co se lépe čte. V tomto programu není dokonale jasné, které všechny případy vlastně `else` pokrývá.

```python
raining = False
have_umbrella = True

if not raining:
    print("Jdu ven.")
elif raining and have_umbrella:
    print("Jdu ven.")
else:
    print("Zůstávám doma.")
```
