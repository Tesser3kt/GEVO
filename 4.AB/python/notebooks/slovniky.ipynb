# Slovníky v Pythonu

Seznamy umožňují uchovávat více objektů na jednom místě, ale k těm objektům se můžeme dostat jen podle jejich pořadí, což není vždy žádoucí. Představme si situaci, kdy potřebujeme ve svém programu uložit osobní údaje mnoha lidí. Pochopitelně, všichni lidé mohou ležet v jednom dlouhém seznamu, ale jejich osobní údaje už ukládat do seznamu není ideální -- museli bychom si pamatovat, že třeba ulice je vždycky na pátém místě, telefon na osmém apod. Takový seznam by mohl vypadat třeba takto:

```python
customer_data = [
    ["Bedřich", "Kanec", "muž", "Celetná", "203/16", "Praha", "kancovna@gmail.com", "773222111"],
    ["Karel", "Servác", "muž", "Wenzigova", "111/3", "Praha", "niesubonifac@seznam.cz", "331892002"]
    ["Pankratěvna", "Praskovna", "žena", "Pařížská", "221/9", "Praha", "dochaziminapady@email.ru", "789123009"]
]
```

Mnohem přehlednější je použít slovník -- datový typ `dict` -- který plní podobný účel jako seznam, ale prvky do něj lze ukládat i pod jiné datové typy než celá čísla, například pod stringy. Slovníky jsou v tomto smyslu více obecné než seznamy, ale tím, že v nich prvky nejsou uspořádané (nemají pořadí), nemůžeme slovníky například řezat ani se ptát na otázky jako: "Který prvek je první?"

Slovníky píšeme do složených závorek `{}` a prvky vkládáme stylem `klíč : hodnota`, kde `klíč` je objekt (obvykle string, ale ne nutně), pod kterým je ve slovníku uložen objekt `hodnota` (jakýkoliv datový typ). Upravený seznam zákazníků s použitím slovníků by mohl vypadat takto:

```python
customer_data = [
    {
        "first_name": "Bedřich",
        "last_name": "Kanec",
        "gender": "muž",
        "street": "Celetná",
        "house_number": "203/16",
        "city": "Praha",
        "email": "kancovna@gmail.com",
        "phone": "773222111 "
    },
    {
        "first_name": "Karel",
        "last_name": "Servác",
        "gender": "muž",
        "street": "Wenzigova",
        "house_number": "111/3",
        "city": "Praha",
        "email": "niesubonifac@seznam.cz",
        "phone": "331892002"
    },
    {
        "first_name": "Pankratěvna",
        "last_name": "Praskovna",
        "gender": "žena",
        "street": "Pařížská",
        "house_number": "221/9",
        "city": "Praha",
        "email": "dochaziminapady@email.ru",
        "phone": "789123009"
    }
]
```

Ze slovníku vybíráme údaje stejným způsobem jako ze seznamu: do hranatých závorek vpíšeme klíč, pod kterým je uložená hodnota. Například:

```python
customer = customer_data[0]
customer["email"]
```

Též jsme mohli napsat jednodušeji `customer_data[0]["email"]`.

Pochopitelně, hodnotou ve slovníku může být i seznam. Je normální, že lidé mají více telefonních čísel. Jeden záznam v seznamu `customer_data` by též mohl vypadat takto, kde je pod klíčem `phone_numbers` uložen seznam telefonních čísel:

```python
{
    "first_name": "Pankratěvna",
    "last_name": "Praskovna",
    "gender": "žena",
    "street": "Pařížská",
    "house_number": "221/9",
    "city": "Praha",
    "email": "dochaziminapady@email.ru",
    "phone_numbers": ["789123009", "222333444"]
}
```

## Shrnutí

Slovníky nám umožňují "pojmenovávat" prvky oproti seznamům, kde každý prvek má pouze dané pořadí. Prvky ve slovníku jsou uložené stylem `klíč : hodnota`, kde klíčem mohou být různé (ale ne všechny) datové typy: mimo jiné `int`, `float` a `str`. K prvkům slovníku se dostáváme stejným způsobem jako k prvkům seznamu: přes `[]`, do nichž vpíšeme klíč. Oproti seznamům nemají prvky slovníku dané pořadí, slovníky nelze řezat, ani slučovat nebo násobit.
