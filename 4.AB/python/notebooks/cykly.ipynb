# Cykly v Pythonu

V programech často narážíme na situace, kdy musíme tutéž operaci opakovat několikrát:
- přidat předvolbu do seznamu čísel,
- spočítat celkovou cenu všech položek v košíku,
- hýbat postavičkou, dokud uživatel drží tlačítko,
- a spousta dalších...

K opakování operací slouží v Pythonu tzv. *cykly* (nebo *loopy*). Jsou dvojího typu:
- `for` cyklus: provede danou operaci (nebo posloupnost operací) na každý prvek z posloupnosti (třeba `str`, `list` nebo `dict`).
- `while` cyklus: provádí danou operaci (nebo posloupnost operací) tak dlouho, dokud platí nějaká podmínka.

## `for` cyklus

Píše se stylem `for element in sequence`, kde `sequence` je nějakého datového typu, který v sobě nese prvky -- my jsme zatím viděli `str`, `list` a `dict`. Proměnná `element` je pak pojmenování pro jeden prvek z této posloupnosti, na který provádíme danou operaci. Jednoduchý příklad níže. Tento `for` cyklus projde seznamem čísel `numbers` a každé jeho číslo (proměnnou `number`) vytiskne vynásobené `5`.

```python
numbers = [4, 1, 8, 9, 1]
for number in numbers:
    print(number * 5)
```

Použitý na stringy funguje `for` cyklus tak, že je prochází symbol po symbolu. V následujícím příkladě spočítáme počet samohlásek ve slově (bez diakritiky).

```python
vowels = ["a", "e", "i", "o", "u", "y"]
word = "Mezopotamie"

vowel_count = 0
for letter in word:
    if letter in vowels:
        vowel_count = vowel_count + 1

print(vowel_count)
```

Vysvětlení:
1. Seznam `vowels` obsahuje české samohlásky (bez diakritiky) a v proměnné `word` je uloženo slovo, jehož počet samohlásek chceme určit.
2. Počet zatím nalezených samohlásek si musíme pojmenovat; od toho je proměnná `vowel_count`, která je z počátku `0`, protože jsme zatím žádnou samohlásku nenašli.
3. Slovo `word` procházíme písmeno po písmeně (proměnná `letter`),  a když je písmeno samohláska (`if letter in vowels`), tak zvýšíme počet `vowel_count` o `1`, neboli uložíme do proměnné `vowel_count` to, co tam bylo `+ 1` (což zařídí `vowel_count = vowel_count + 1`).

Ukážeme příklad programu, který přidává předvolby k telefonním číslům podle země. Budeme mít slovník s předvolbami a seznam lidí s telefonními čísly a zeměmi původu.

```python
prefixes = {
    "Germany": "+49",
    "Slovakia": "+421",
    "Norway": "+47"
}

people = [
    {
        "name": "Johann Sebastian Bach",
        "city": "Eisenach",
        "country": "Germany",
        "phone": "331920222"
    },
    {
        "name": "Robert Fico",
        "city": "Topoľčany",
        "country": "Slovakia",
        "phone": "118223091"
    },
    {
        "name": "Magnus Carlsen",
        "city": "Tønsberg",
        "country": "Norway",
        "phone": "33722822"
    }
]

for person in people:
    phone_prefix = prefixes[person["country"]]
    person["phone"] = phone_prefix + person["phone"]

print(people)
```

Vysvětlení:
1. Procházíme seznam `people` člověk po člověku (proměnná `person`).
2. U každého člověka použijeme jeho zemi původu `person["country"]` jako klíč do slovníku `prefixes`, z něhož vytáhneme správnou předvolbu.
3. Do položky `person["phone"]` každého člověka uložíme tu samou hodnotu akorát s předvolbou.

`for` cyklus použitý na slovník (`dict`) jej prochází po klíčích. Kdybychom chtěli z předchozího příkladu jen vytisknout všechny země s jejich telefonními předvolbami, mohli bychom napsat například toto.

```python
for country in prefixes:
    print("The telephone prefix of", country, "is", prefixes[country])
```

### Funkce `range`

Často se nám stane, že potřebujeme nějakou akci opakovat zkrátka desetkrát a ne nezbytně vybírat prvky z nějaké posloupnosti. Python má pro toto řešení v podobě funkce `range`, která dostává tři údaje: `start`, `end` a `range` a vyrobí seznam čísel počínaje `start` (včetně) a konče `end` (mimo), která jsou od sebe vzdálena `range`. Vlastně má úplně stejnou syntaxi jako řezy seznamů nebo stringů. Třeba `range(3, 10, 2)` vytvoří `[3, 5, 7, 9]` -- seznam čísel od `3` do `10` se vzdáleností `2`.

```python
for number in range(3, 10, 2):
    print(number)
```

Parametry `start` a `step` nejsou povinné ve funkci `range`. Můžeme zkrátka napsat `range(5)`, čímž získáme seznam `[0, 1, 2, 3, 4]`, čili seznam prvních pěti čísel. Povšimněme si taky, že čísla `0, 1, 2, 3, 4` jsou **indexy prvních pěti prvků seznamu nebo stringu**. Kdybych chtěl tedy například vypsat prvních pět symbolů z nějakého stringu `string`, mohl bych tak učinit pomocí řezů přes `string[:5]` nebo též pomocí indexů přes `range(5)`. Dole vidíte obě možnosti.

```python
string = "whatever"

for symbol in string[:5]:
    print(symbol)

for index in range(5):
    print(string[index])
```

## `while` cyklus

V případě, že potřebujeme opakovat nějakou akci tak dlouho, dokud platí jistá podmínka, je nám k disposici `while` cyklus. Jako příklad uvažme situaci, kdy dostaneme mocninu dvojky a máme odpovědět, kolikátá mocnina to je. Zde nám pochopitelně žádný seznam ani funkce `range` nepomohou, jelikož **nevíme, kolikrát budeme muset číslo dělit dvěma**. Velmi hrubě řečeno platí, že `for` cykly jsou užitečné ve chvíli, kdy předem známe počet opakování a `while` cykly naopak ve chvíli, kdy počet opakování neznáme.

```python
power_of_two = 65536
power = 0

while power_of_two > 1:
    power_of_two = power_of_two // 2
    power += 1

print(power)
```

Vysvětlení:
1. Chceme určit, kolikátá mocnina dvojky je číslo `power_of_two`.
2. Budeme tedy počítat, kolikrát je dokážeme vydělit dvěma, než se dostaneme na `1`. K tomu slouží `while power_of_two > 1`, cyklus, který poběží tak dlouho, dokud je číslo `power_of_two` větší než `1`.
3. Do proměnné `power_of_two` uložíme totéž číslo vydělené dvěma a k proměnné power přičteme `1` (výraz `power += 1` je zkratkou za `power = power + 1`).

Jako další příklad uvažme program, který otravuje uživatele tak dlouho, dokud netvrdí, že jeho oblíbené jídlo je *utopenec*.

```python
fav_food = input("Zadejte své oblíbené jídlo: ")
while fav_food != "utopenec":
    print("Co jste to za prase? Zkuste to znovu.")
    fav_food = input("Zadejte své oblíbené jídlo: ")

print("Tak je to správně.")
```

**Varování**: pomocí `while` cyklů lze snadno psát **nekonečné** programy (třeba přes `while True`). Jako programátorů je vaší prací zajistit, že napsaný `while` cyklus někdy skončí.

### Shrnutí

Python obsahuje dva typy cyklů: `for` a `while`. Cyklus `for` slouží k provádění téže operace s každým prvkem posloupnosti objektů. Cyklus `while` slouží k opakování téže operace tak dlouho, dokud je splněna jistá podmínka.
