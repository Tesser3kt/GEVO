# Stringy v Pythonu

Slova, věty a obecně posloupnosti symbolů můžeme v Pythonu ukládat pomocí tzv. *stringů*. Stringy **vždy píšeme do uvozovek** (ať už jednoduchých `'` nebo dvojitých `"`). Tedy, například `"letadlo"`, `"Mám rád kočky."` nebo `"qwšěěídí393D3M('S'A)"` jsou stringy, ale třeba `32289` nebo `ddmkmš` (bez uvozovek) ne.

#prvního # Operace se stringy

Stringy můžeme
- slučovat: `"kočka" + "pes"` dá v Pythonu string `"kočkapes"`.
	- **mezera** `" "` **je** v Pythonu **též symbol**, takže můžeme slučovat stringy také takto: `"Chci" + " " + "domů."`, což Python spočítá jako `"Chci domů."`
- opakovat: `"kočka" * 3` Python chápe jako `"kočkakočkakočka"`. Zkrátka, násobení stringu celým číslem ho zopakuje tolikrát, kolik je to číslo.
- indexovat: `"kočka"[3]` vybere *čtvrté* písmeno ze stringu `"kočka"`, tedy `"k"`.
	- **Pozor!** Pro Python **je prvním přirozeným číslem 0**. Takže první písmeno ve stringu má index 0, druhé má index 1 atd.
	- Indexem vždycky **musí být celé číslo**, ale může být i *záporné*. Třeba `"kočka"[-1]` vybere *první písmeno zezadu* ze stringu `"kočka"`, tedy `"a"`.
	- Vybraná písmena jsou samy o sobě stringy, takže můžeme v Pythonu napsat věci jako
		- `"kočka"[1] + "kočka"[-2]`, což spojí druhé písmeno zleva s druhým písmenem zprava ze stringu `"kočka"` a vyrobí string `"ok"`;
		- `"kočka[3] * 5"`, což pětkrát zopakuje čtvrté písmeno ze stringu `"kočka"` a vyrobí string `"kkkkk"`.
- řezat/slicovat: `"kočka"[1:4]` vybere všechna písmena ze stringu `"kočka"` **od druhého po čtvrté**, tedy vyrobí string `"očk"`.
	- **Pozor!** Řezy/slicy berou od prvního indexu **včetně** do druhého indexu **mimo**, takže třeba zmíněné `"kočka"[1:4]` vybere písmena v intervalu `<1:4)`: písmeno s indexem `1` ano, písmeno s indexem `4` už ne.
	- Při řezání/slicování *můžete libovolný z obou indexů vynechat* a bere se jako první/poslední:
		- `"kočka"[:3]` vybere všechna písmena od počátečního až po třetí a vyrobí string `"koč"`;
		- `"kočka"[1:]` vybere všechna písmena od druhého až po poslední a vyrobí string `"očka"`;
		- záporné indexy fungují také: třeba `"kočka"[-3:-1]"` vybere písmena od třetího zprava *včetně* do prvního zprava *mimo* a vyrobí string `"čk"`;
		- můžete též specifikovat po kolika písmenech má Python při řezání skákat: například `"popocatepetl"[1:-2:2]` vybere **každé druhé písmeno** od písmene s indexem `1` včetně do písmene s indexem `-2` mimo a vyrobí string `"ooaee"`.
		- obecně je syntaxe řezů/sliců tato: `"string"[levý index:pravý index:krok]`.

Co se stane, když se pokusím vybrat šesté písmeno ze slova `"kočka"`? Zamyslete se a pak to vyzkoušejte.

```python
"kočka"[5]
```

Ve řezu/slicu můžeme kterékoli z čísel vynechat (levý index, pravý index nebo krok). Když napíšu `"kočka"[::-1]"` říkám Pythonu: "Vyřízni ze stringu `"kočka"` symboly od počátečního indexu (vynechán) do posledního indexu (vynechán) a skákej po `-1` symbolech." Co to znamená skákat po `-1` symbolech? Co se asi stane? Zkuste si to rozmyslet, než kód spustíte.

```python
"kočka"[::-1]
```

## Shrnutí

Stringy se v Pythonu používají k práci s libovolnými posloupnostmi symbolů, často slovy a větami. Stringy můžeme **slučovat** a **opakovat** a **vyřezávat** z nich jednotlivé symboly nebo i kusy symbolů.
