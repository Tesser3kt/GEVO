# Seznamy v Pythonu

Čísla ani stringy nám neumožňují uchovávat více různých objektů na jedno místě. Od toho jsou v Pythonu *seznamy* -- datový typ `list` -- které slouží přesně tomuto účelu. Seznamy mají prvky, které můžeme "vytáhnout" pomocí jejich pořadí v seznamu, podobně jako symboly ze stringů.

Seznamy se píší do hranatých závorek `[]`, do kterých vpisujeme jejich prvky oddělené čárkami: např. `[1, 5, 7, "kocka"]`. Prvky uložené v seznamu mají své pořadí (též *index*), podle kterého se k nim můžeme v seznamu dostat.

```python
numbers = [5, 4.9, 1, -3, -8.5, 3.1]
numbers[2] + numbers[-2]
```

V příkladu výše jsem do proměnné `numbers` uložil seznam `[5, 4.9, 1, -3, -8.5, 3.1]` a poté jsem sečetl jeho třetí prvek zleva s jeho druhým prvkem zprava.

Prvky seznamů mohou být jakéhokoliv datového typu: čísla, stringy, ale klidně i *další seznamy*. Jako seznamy seznamů můžeme v Pythonu representovat třeba tabulky.

```python
table = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

table[1][2]
```

K prvkům seznamu seznamů se dostaneme přes dvojité indexování. Nejprve si ze seznamu table vytáhneme prvek `table[1]`, což je seznam `[4, 5, 6]` a pak z tohoto seznamu vytáhneme prvek s indexem `2`, což je číslo `6`.

Seznamy můžeme řezat stejně jako stringy. Rozmyslete si, co se stane v buňkách níže.

```python
words = ["auto", "hotel", "kráva", "sklenice"]
words[1:-1:1]
```

```python
table = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

table[2:][-1][:-1]
```

Stejně jako stringy, lze i seznamy slučovat a opakovat. Například
- `[5, 4] + [4.7, 2.3]` vyrobí seznam `[5, 4, 4.7, 2.3]`.
- `[3, 4] * 3` vyrobí seznam `[3, 4, 3, 4, 3, 4]`.

## Shrnutí

Seznamy v Pythonu jsou datové typy pro ukládání více objektů na jednom místě. Uložené objekty mohou být jakéhokoliv datového typu a mají jednoznačně pořadí v seznamu. Pomocí tohoto pořadí je se seznamu vytahujeme. Seznamy můžeme, stejně jako stringy, řezat, slučovat a opakovat.
